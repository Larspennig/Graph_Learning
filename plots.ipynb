```python
import torch
import numpy as np
import pyvista as pv
import torch_geometric as tg
import open3d as o3d
from loaders.Sndataloader2 import SNpart_Dataset
```

```python
dataset_train = SNpart_Dataset(root='data/shapenet_core/',
                                   split='train')
```

```python
data = dataset_train[4]
```

```python
points = data.pos.numpy()
normals = data.x.numpy()[:,:3]
labels = data.y.numpy()
```

```python
color = [1, 1, 1]  # Red

# Step 3: Paint all points with the same color
# Get the number of points in the point cloud
num_points = len(points)

# Create a color array where each point has the same color
colors = np.tile(color, (num_points, 1))

# Assign this color to the point cloud
```

```python
# ploting normals
cloudp = o3d.geometry.PointCloud(o3d.utility.Vector3dVector(points))
cloudp.normals = o3d.utility.Vector3dVector(o3d.utility.Vector3dVector(normals))
cloudp.colors = o3d.utility.Vector3dVector(colors)

o3d.visualization.draw_geometries([cloudp], point_show_normal=True)
```

```python
# plotting part segementation
color = np.zeros((len(labels),3))
color[labels == 12] = [100,0,0]
color[labels == 13] = [0,100,0]
color[labels == 14] = [0,0,100]
color[labels == 15] = [100,100,0]
color[labels == 35] = [10,10,10]

cloud = pv.PolyData(points)
cloud['colors'] = color


pv.plot(cloud, render_points_as_spheres = True, point_size = 8, cmap = 'YlGnBu',rgb=True)
```
